CONTOH PENERAPAN "LIST"

```python
penjualan_harian = [100, 150, 200, 180, 220, 250, 300]
print(penjualan_harian[3])  # Output: 180
print(penjualan_harian[-1])  # Output: 300
print(len(penjualan_harian))  # Output: 7
```

CONTOH PENERAPAN "DICTIONARY"

```python
profil_toko = {
    "nama": "Toko ABC",
    "alamat": "Jl. Contoh No. 123",
    "telepon": "08123456789",
    "email": "tokoabc@gmail.com"
}

print(profil_toko["nama"])  # Output: Toko ABC
print(profil_toko["alamat"])  # Output: Jl. Contoh No. 123
print(profil_toko["telepon"])  # Output: 08123456789
print(profil_toko["email"])  # Output: tokoabc@gmail.com
```

CONTOH PENERAPAN "TUPPLE"

```python
koordinat_toko = (-6.9147, 107.6098) # (lintang, bujur), tuple ditulis dengan kurung biasa
print(koordinat_toko[0]) # -6.9147
# Baris berikut akan memicu error kalau dijalankan:

try:
    koordinat_toko[0] = -6.9000 # tuple bersifat immutable, tidak bisa diubah
except Exception as e:
    print(f"{type(e).__name__}: {e}")
```

CONTOH PENERAPAN "SET"

```python
kota_operasional = {"Bandung", "Jakarta", "Surabaya", "Bandung"}
print(kota_operasional) # {"Bandung", "Jakarta", "Surabaya"} -> duplikat otomatis hilang
print(len(kota_operasional)) # 3, bukan 4
kota_a = {"Bandung", "Jakarta", "Surabaya"}
kota_b = {"Jakarta", "Semarang"}
print(kota_a & kota_b) # irisan: {"Jakarta"}
print(kota_a | kota_b) # gabungan: {"Bandung", "Jakarta", "Surabaya", "Semarang"}
```
