CONTOH PENERAPAN "SET"

```python
kota_operasional = {"Bandung", "Jakarta", "Surabaya", "Bandung"}
print(kota_operasional) # {"Bandung", "Jakarta", "Surabaya"} -> duplikat otomatis hilang
print(len(kota_operasional)) # 3, bukan 4
kota_a = {"Bandung", "Jakarta", "Surabaya"}
kota_b = {"Jakarta", "Semarang"}
print(kota_a & kota_b) # irisan: {"Jakarta"}
print(kota_a | kota_b) # gabungan: {"Bandung", "Jakarta", "Surabaya", "Semarang"}
```

CONTOH PENERAPAN "LOOPING"

```python
penjualan_harian1 = [120, 95, 140, 88, 175]
total = 0

for angka in penjualan_harian1:
    total = total + angka
print(total) # 618

# Pola yang sama dengan while, sekadar untuk menunjukkan perbedaannya:
i = 0
total_while = 0
while i < len(penjualan_harian1):
    total_while = total_while + penjualan_harian1[i]
    i = i + 1

print(total_while) # 618, hasil identik
```

CONTOH PENERAPAN "LINEAR SEARCH"

```python
daftar_toko_list = ["Toko A", "Toko B", "Toko C"]
target_toko = "Toko B"

ditemukan = False
for toko in daftar_toko_list:
    if toko == target_toko:
        ditemukan = True
        break

print(ditemukan) # True, karena "Toko B" ada di daftar_toko_list
```

CONTOH PENERAPAN "LINEAR SEARCH DENGAN BIG DATA"

```python
daftar_besar = [f"Pelanggan-{i:07d}" for i in range(1_200_000)]
target = "Pelanggan-1199999" # posisi terakhir, skenario terburuk

def linear_search_hitung(data, target):
    perbandingan = 0
    for item in data:
        perbandingan += 1
        if item == target:
            return perbandingan
    return perbandingan

jumlah = linear_search_hitung(daftar_besar, target)
print(f"Linear search (list): {jumlah:,} perbandingan")

set_besar = set(daftar_besar)
print(f"Pengecekan via set: ditemukan={target in set_besar}")
```

CONTOH PENERAPAN "SORTING (NORMAL METHOD)"

```python
penjualan_harian = [120, 95, 140, 88, 175]

print(sorted(penjualan_harian)) # [88, 95, 120, 140, 175]
print(sorted(penjualan_harian, reverse=True)) # [175, 140, 120, 95, 88]

def bubble_sort_manual(data):
    data = data.copy() # supaya list aslinya tidak ikut berubah
    n = len(data)
    for putaran in range(n):
        for i in range(n - putaran - 1):
            if data[i] > data[i + 1]:
                data[i], data[i + 1] = data[i + 1], data[i]
    return data

print(bubble_sort_manual(penjualan_harian)) # [88, 95, 120, 140, 175]
```

WORKED EXAMPLE

```python
transaksi_mentah = [
    {"toko": "Toko Melati", "kota": "Bandung", "total": 87000},
    {"toko": "Toko Kenanga", "kota": "Jakarta", "total": 152000},
    {"toko": "Toko Anggrek", "kota": "Bandung", "total": 63000},
    {"toko": "Toko Mawar", "kota": "Surabaya", "total": 210000},
    {"toko": "Toko Melati", "kota": "Bandung", "total": 94000},
]

transaksi_bandung = []

for transaksi in transaksi_mentah:
    if transaksi["kota"] == "Bandung":
        transaksi_bandung.append(transaksi["total"])

transaksi_bandung_terurut = sorted(transaksi_bandung, reverse=True)

print(transaksi_bandung_terurut) # [94000, 87000, 63000]
print(sum(transaksi_bandung_terurut)) # 244000
```

SOAL 1

```python
nilai_ujian = [78, 92, 65, 88, 71, 95, 60]

tertinggi = nilai_ujian[0]
terendah = nilai_ujian[0]
total = 0
for nilai in nilai_ujian:
    if nilai > tertinggi:
        tertinggi = nilai
    if nilai < terendah:
        terendah = nilai
    total = total + nilai

rata_rata = total / len(nilai_ujian)

print(f"Tertinggi: {tertinggi}, Terendah: {terendah}, Rata-rata: {rata_rata:.2f}")
```
